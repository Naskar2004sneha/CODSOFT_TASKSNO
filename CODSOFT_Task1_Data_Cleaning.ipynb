```python
import pandas as pd

file_path = "/content/adult.data"

df = pd.read_csv(file_path, header=None, skipinitialspace=True)

print("Dataset loaded successfully!")
print("First 5 rows:")
display(df.head())
```

```python
column_names = [
    "age", "workclass", "fnlwgt", "education",
    "education_num", "marital_status", "occupation",
    "relationship", "race", "sex", "capital_gain",
    "capital_loss", "hours_per_week", "native_country",
    "income"
]

df.columns = column_names

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
# Count question marks in each column
missing_counts = (df == "?").sum()

print("Missing entries marked with '?':")
print(missing_counts[missing_counts > 0])

print("\nTotal missing entries:", missing_counts.sum())
```

```python
# Replace '?' with proper missing values
df.replace("?", pd.NA, inplace=True)

# Count missing values after replacement
print("Missing values after replacement:")
print(df.isnull().sum())

# Remove rows containing missing values
df_cleaned = df.dropna().copy()

print("\nOriginal dataset shape:", df.shape)
print("Cleaned dataset shape:", df_cleaned.shape)
print("Rows removed:", len(df) - len(df_cleaned))
```

```python
# Count duplicate rows
duplicate_count = df_cleaned.duplicated().sum()

print("Duplicate rows found:", duplicate_count)

# Remove duplicate rows
df_cleaned = df_cleaned.drop_duplicates().copy()

print("Shape after removing duplicates:", df_cleaned.shape)
print("Remaining duplicate rows:", df_cleaned.duplicated().sum())
```

```python
print("Data types before correction:")
print(df_cleaned.dtypes)

# Convert numerical columns to numeric types
numeric_columns = [
    "age", "fnlwgt", "education_num",
    "capital_gain", "capital_loss", "hours_per_week"
]

for column in numeric_columns:
    df_cleaned[column] = pd.to_numeric(
        df_cleaned[column], errors="coerce"
    )

print("\nData types after correction:")
print(df_cleaned.dtypes)
```

```python
# Remove unnecessary spaces from text columns
text_columns = df_cleaned.select_dtypes(
    include="object"
).columns

for column in text_columns:
    df_cleaned[column] = df_cleaned[column].str.strip()

# Inspect unique values in important columns
print("Workclass categories:")
print(df_cleaned["workclass"].unique())

print("\nOccupation categories:")
print(df_cleaned["occupation"].unique())

print("\nIncome categories:")
print(df_cleaned["income"].unique())
```

```python
print("Final dataset shape:", df_cleaned.shape)

print("\nMissing values remaining:")
print(df_cleaned.isnull().sum().sum())

print("\nDuplicate rows remaining:")
print(df_cleaned.duplicated().sum())

print("\nData types:")
print(df_cleaned.dtypes)
```

```python
# Save the cleaned dataset as a CSV file
output_file = "/content/cleaned_adult_data.csv"

df_cleaned.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File location:", output_file)
```
